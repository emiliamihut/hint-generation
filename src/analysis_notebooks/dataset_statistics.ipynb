# Track 0 â€” Dataset Statistics


```python
from pathlib import Path
import json
import re
import unicodedata

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

PROJECT = Path(r"D:\Hint-generation\hackerrank_scraper")
DATA = PROJECT / "data"

RAW_WITH_CODE = [
    DATA / "raw_original" / "submissions_with_code.jsonl",
    DATA / "raw" / "submissions_with_code.jsonl",
    DATA / "submissions_with_code.jsonl",
]

RAW_METADATA = [
    DATA / "raw_original" / "submissions_metadata.jsonl",
    DATA / "raw" / "submissions_metadata.jsonl",
    DATA / "submissions_metadata.jsonl",
]

CLEAN_ROOT = DATA / "clean_solutions" / "submissions"
ALGORITHM_FILE = DATA / "algorithm_analysis" / "hybrid_algorithm_labels.csv"

OUTPUT = DATA / "algorithm_analysis" / "track0_statistics"
OUTPUT.mkdir(parents=True, exist_ok=True)

PROBLEMS = [
    "matei-si-diamantele",
    "gigel-si-raspandirea-vestilor",
    "gigel-si-sosetele",
    "gigel-si-telefonul",
    "organizare-mafiota",
    "cutiile-magice",
]
```

```python
def first_existing(paths):
    for path in paths:
        if path.exists():
            return path
    return None


def load_jsonl(path):
    rows = []
    with path.open("r", encoding="utf-8-sig", errors="replace") as f:
        for line in f:
            line = line.strip()
            if line:
                rows.append(json.loads(line))
    return pd.DataFrame(rows)


raw_code_path = first_existing(RAW_WITH_CODE)
raw_meta_path = first_existing(RAW_METADATA)

if raw_code_path is None and raw_meta_path is None:
    raise FileNotFoundError("Raw submission files were not found.")

if raw_code_path is not None:
    raw = load_jsonl(raw_code_path)
else:
    raw = load_jsonl(raw_meta_path)
```

```python
def pick_column(df, names, required=False):
    for name in names:
        if name in df.columns:
            return name
    if required:
        raise KeyError(f"Missing column. Tried: {names}")
    return None


def normalize_problem(value):
    text = unicodedata.normalize("NFKD", str(value))
    text = "".join(c for c in text if not unicodedata.combining(c))
    text = text.lower().replace("_", "-")
    text = re.sub(r"[^a-z0-9]+", "-", text).strip("-")

    for problem in PROBLEMS:
        if problem in text:
            return problem

    return text


def normalize_language(value):
    text = str(value).strip().lower()

    if "java" in text:
        return "Java"
    if "cpp" in text or "c++" in text or "gnu++" in text:
        return "C++"
    if text == "c" or text.startswith("c99") or text.startswith("c11") or text.startswith("gnu c"):
        return "C"

    return str(value).strip()


problem_col = pick_column(raw, ["problem", "problem_slug", "challenge", "challenge_slug"], required=True)
submission_col = pick_column(raw, ["submission_id", "id", "submissionId"], required=True)
user_col = pick_column(raw, ["username", "user_id", "user", "team", "team_name"], required=True)
score_col = pick_column(raw, ["score", "points", "score_site"], required=True)
language_col = pick_column(raw, ["language", "lang", "language_name"], required=True)
time_col = pick_column(raw, ["timestamp", "time", "submitted_at", "created_at", "date"])
code_col = pick_column(raw, ["code", "source", "source_code", "submission_code"])

df = pd.DataFrame({
    "problem": raw[problem_col].map(normalize_problem),
    "submission_id": raw[submission_col].astype(str),
    "user_id": raw[user_col].astype(str),
    "score": pd.to_numeric(raw[score_col], errors="coerce"),
    "language": raw[language_col].map(normalize_language),
})

df["timestamp"] = raw[time_col] if time_col is not None else ""
df["code"] = raw[code_col].fillna("").astype(str) if code_col is not None else ""

df = df[df["problem"].isin(PROBLEMS)].copy()
df = df.dropna(subset=["score"])

display(df.head())
```

## Dataset Overview


```python
def count_clean_solutions(problem):
    folder = CLEAN_ROOT / problem
    if not folder.exists():
        return 0

    valid_extensions = {".c", ".cc", ".cpp", ".cxx", ".java"}

    return sum(
        1
        for path in folder.iterdir()
        if path.is_file() and path.suffix.lower() in valid_extensions
    )


overview = (
    df.groupby("problem", observed=True)
      .agg(
          raw_submissions=("submission_id", "count"),
          users=("user_id", "nunique"),
      )
      .reindex(PROBLEMS)
)

overview["clean_solutions"] = [
    count_clean_solutions(problem)
    for problem in overview.index
]

overview = overview.reset_index()

total_row = pd.DataFrame([{
    "problem": "TOTAL",
    "raw_submissions": int(overview["raw_submissions"].sum()),
    "users": int(df["user_id"].nunique()),
    "clean_solutions": int(overview["clean_solutions"].sum()),
}])

overview_with_total = pd.concat([overview, total_row], ignore_index=True)

display(overview_with_total)
overview_with_total.to_csv(OUTPUT / "dataset_overview.csv", index=False)
```

## Score Summary


```python
score_stats = (
    df.groupby("problem", observed=True)
      .agg(
          submissions=("submission_id", "count"),
          min_score=("score", "min"),
          max_score=("score", "max"),
          mean_score=("score", "mean"),
          median_score=("score", "median"),
      )
      .reindex(PROBLEMS)
      .round(2)
      .reset_index()
)

display(score_stats)
score_stats.to_csv(OUTPUT / "score_statistics.csv", index=False)
```

## Score Distribution


```python
score_distribution = (
    df.groupby(["problem", "score"], observed=True)
      .size()
      .rename("submissions")
      .reset_index()
      .sort_values(["problem", "score"])
)

display(score_distribution)

for problem in PROBLEMS:
    part = score_distribution[score_distribution["problem"] == problem]

    if part.empty:
        continue

    plt.figure(figsize=(8, 4))
    plt.bar(part["score"].astype(str), part["submissions"])
    plt.title(f"Submissions by Score â€” {problem}")
    plt.xlabel("Score")
    plt.ylabel("Submissions")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

score_distribution.to_csv(OUTPUT / "score_distribution.csv", index=False)
```

## Language Distribution


```python
language_distribution = (
    df.groupby(["problem", "language"], observed=True)
      .size()
      .rename("submissions")
      .reset_index()
)

display(language_distribution)

for problem in PROBLEMS:
    part = language_distribution[language_distribution["problem"] == problem]

    if part.empty:
        continue

    plt.figure(figsize=(6, 4))
    plt.bar(part["language"], part["submissions"])
    plt.title(f"Submissions by Language â€” {problem}")
    plt.xlabel("Language")
    plt.ylabel("Submissions")
    plt.tight_layout()
    plt.show()

language_distribution.to_csv(OUTPUT / "language_distribution.csv", index=False)
```

## Code Length


```python
def non_empty_lines(code):
    return sum(
        1
        for line in str(code).splitlines()
        if line.strip()
    )


df["submission_length"] = df["code"].map(non_empty_lines)

length_stats = (
    df.groupby("problem", observed=True)
      .agg(
          mean_lines=("submission_length", "mean"),
          median_lines=("submission_length", "median"),
          min_lines=("submission_length", "min"),
          max_lines=("submission_length", "max"),
      )
      .reindex(PROBLEMS)
      .round(2)
      .reset_index()
)

display(length_stats)

for problem in PROBLEMS:
    part = df[df["problem"] == problem]

    plt.figure(figsize=(7, 4))
    plt.hist(part["submission_length"], bins=20)
    plt.title(f"Code Length Distribution â€” {problem}")
    plt.xlabel("Non-empty lines")
    plt.ylabel("Submissions")
    plt.tight_layout()
    plt.show()

length_stats.to_csv(OUTPUT / "submission_length.csv", index=False)
```

## Submission Traces


```python
trace_sizes = (
    df.groupby(["problem", "user_id"], observed=True)
      .size()
      .rename("trace_length")
      .reset_index()
)

trace_summary = (
    trace_sizes.groupby("problem", observed=True)
               .agg(
                   traces=("user_id", "count"),
                   users_with_more_than_one_submission=(
                       "trace_length",
                       lambda x: int((x > 1).sum())
                   ),
                   mean_trace_length=("trace_length", "mean"),
                   max_trace_length=("trace_length", "max"),
               )
               .reindex(PROBLEMS)
               .round(2)
               .reset_index()
)

display(trace_summary)

plt.figure(figsize=(9, 4))
plt.bar(trace_summary["problem"], trace_summary["traces"])
plt.title("User Traces by Problem")
plt.xlabel("Problem")
plt.ylabel("Traces")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()

trace_sizes.to_csv(OUTPUT / "submission_traces.csv", index=False)
trace_summary.to_csv(OUTPUT / "submission_trace_summary.csv", index=False)
```

## Attempts per User


```python
for problem in PROBLEMS:
    part = trace_sizes[trace_sizes["problem"] == problem]

    if part.empty:
        continue

    counts = (
        part["trace_length"]
        .value_counts()
        .sort_index()
    )

    plt.figure(figsize=(7, 4))
    plt.bar(counts.index.astype(str), counts.values)
    plt.title(f"Attempts per User â€” {problem}")
    plt.xlabel("Trace length")
    plt.ylabel("Users")
    plt.tight_layout()
    plt.show()
```

## Predicted Algorithms


```python
if not ALGORITHM_FILE.exists():
    raise FileNotFoundError(f"Missing file: {ALGORITHM_FILE}")

alg = pd.read_csv(ALGORITHM_FILE)

if "predicted_algorithm" not in alg.columns:
    raise KeyError("The file does not contain predicted_algorithm.")

alg_problem_col = pick_column(
    alg,
    ["problem", "problem_slug", "challenge", "challenge_slug"],
    required=True
)

alg_score_col = pick_column(
    alg,
    ["score", "score_site", "points"],
    required=True
)

alg["problem"] = alg[alg_problem_col].map(normalize_problem)
alg["score"] = pd.to_numeric(alg[alg_score_col], errors="coerce")
alg["predicted_algorithm"] = alg["predicted_algorithm"].fillna("unknown").astype(str)

alg = alg[alg["problem"].isin(PROBLEMS)].copy()
alg = alg.dropna(subset=["score"])

display(
    alg[["problem", "score", "predicted_algorithm"]].head()
)
```

```python
predicted_algorithm_distribution = (
    alg.groupby(["problem", "predicted_algorithm"], observed=True)
       .size()
       .rename("submissions")
       .reset_index()
       .sort_values(
           ["problem", "submissions"],
           ascending=[True, False]
       )
)

display(predicted_algorithm_distribution)

for problem in PROBLEMS:
    part = predicted_algorithm_distribution[
        predicted_algorithm_distribution["problem"] == problem
    ]

    if part.empty:
        continue

    plt.figure(figsize=(8, 4))
    plt.bar(part["predicted_algorithm"], part["submissions"])
    plt.title(f"Predicted Algorithms â€” {problem}")
    plt.xlabel("Predicted algorithm")
    plt.ylabel("Submissions")
    plt.xticks(rotation=35, ha="right")
    plt.tight_layout()
    plt.show()

predicted_algorithm_distribution.to_csv(
    OUTPUT / "predicted_algorithm_distribution.csv",
    index=False
)
```

## Algorithms by Score


```python
max_score = (
    alg.groupby("problem")["score"]
       .transform("max")
)

alg["score_group"] = np.where(
    alg["score"] == 0,
    "zero",
    np.where(
        alg["score"] >= max_score,
        "max",
        "partial"
    )
)

algorithm_by_score = (
    alg.groupby(
        ["problem", "score_group", "predicted_algorithm"],
        observed=True
    )
    .size()
    .rename("submissions")
    .reset_index()
    .sort_values(
        ["problem", "score_group", "submissions"],
        ascending=[True, True, False]
    )
)

display(algorithm_by_score)

for problem in PROBLEMS:
    part = algorithm_by_score[
        algorithm_by_score["problem"] == problem
    ]

    if part.empty:
        continue

    pivot = part.pivot_table(
        index="predicted_algorithm",
        columns="score_group",
        values="submissions",
        fill_value=0,
        aggfunc="sum"
    )

    ax = pivot.plot(kind="bar", figsize=(9, 4))
    ax.set_title(f"Algorithms by Score Group â€” {problem}")
    ax.set_xlabel("Predicted algorithm")
    ax.set_ylabel("Submissions")
    plt.xticks(rotation=35, ha="right")
    plt.tight_layout()
    plt.show()

algorithm_by_score.to_csv(
    OUTPUT / "predicted_algorithm_by_score.csv",
    index=False
)
```
