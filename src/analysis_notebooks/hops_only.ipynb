```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

ROOT = Path(r"D:\Hint-generation")
TRANSITIONS = ROOT / "hackerrank_scraper" / "data" / "model_preparation" / "trace_transitions.csv"
OUT = ROOT / "MODEL_PREPARATION_SIMPLE"
OUT.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(TRANSITIONS)

PROBLEM_ORDER = [
    "matei-si-diamantele",
    "gigel-si-raspandirea-vestilor",
    "gigel-si-sosetele",
    "gigel-si-telefonul",
    "organizare-mafiota",
    "cutiile-magice",
]

def pick(names):
    m = {str(c).lower(): c for c in df.columns}
    for name in names:
        if name.lower() in m:
            return m[name.lower()]
    return None

problem_col = pick(["problem", "problem_slug"])
from_col = pick([
    "from_score", "score_from", "previous_score", "prev_score",
    "score_before", "before_score", "source_score", "old_score"
])
to_col = pick([
    "to_score", "score_to", "next_score",
    "score_after", "after_score", "target_score", "new_score"
])

if problem_col is None or from_col is None or to_col is None:
    raise ValueError(f"Columns found: {list(df.columns)}")

df["_from"] = pd.to_numeric(df[from_col], errors="coerce").round(2)
df["_to"] = pd.to_numeric(df[to_col], errors="coerce").round(2)
df["_problem"] = df[problem_col].astype(str)

positive = df[df["_to"] > df["_from"]].copy()

def fmt(x):
    if pd.isna(x):
        return ""
    x = float(x)
    if abs(x - round(x)) < 1e-9:
        return str(int(round(x)))
    return f"{x:.2f}".rstrip("0").rstrip(".")
```

```python
hops = (
    positive
    .groupby(["_problem", "_from", "_to"])
    .size()
    .reset_index(name="transitions")
    .rename(columns={
        "_problem": "problem",
        "_from": "from_score",
        "_to": "to_score",
    })
)

hops["gain"] = (hops["to_score"] - hops["from_score"]).round(2)

hops = hops.sort_values(
    ["problem", "transitions", "gain"],
    ascending=[True, False, False]
)

hops.to_csv(
    OUT / "strict_positive_hops.csv",
    index=False
)

display(hops)
```

```python
TOP = 12

for problem in PROBLEM_ORDER:
    part = hops[hops["problem"] == problem].head(TOP).copy()

    if part.empty:
        continue

    part["hop"] = (
        part["from_score"].map(fmt)
        + " â†’ "
        + part["to_score"].map(fmt)
    )

    display(
        part[
            ["from_score", "to_score", "gain", "transitions"]
        ].reset_index(drop=True)
    )

    plot_df = part.iloc[::-1]

    plt.figure(figsize=(10, 5.5))
    bars = plt.barh(plot_df["hop"], plot_df["transitions"])
    plt.title(problem)
    plt.xlabel("Strictly positive transitions")
    plt.ylabel("Score hop")

    for bar, value in zip(bars, plot_df["transitions"]):
        plt.text(
            bar.get_width(),
            bar.get_y() + bar.get_height() / 2,
            f" {int(value)}",
            va="center"
        )

    plt.tight_layout()
    plt.show()
```
